```python
import  pandas as pd
import numpy  as np
```

```python

data = pd.read_csv("../data/credit_risk_dataset.csv")
data_clean = data.copy()

data_clean.head(10)
```

```python
# Renommer  les  colonnes  en français 


colonnes_fr = {
    'person_age': 'age',
    'person_income': 'revenu_annuel',
    'person_home_ownership': 'statut_logement',
    'person_emp_length': 'anciennete_emploi',
    'loan_intent': 'motif_pret',
    'loan_grade': 'niveau_de_risque',
    'loan_amnt': 'montant_pret',
    'loan_int_rate': 'taux_interet',
    'loan_status': 'statut_pret',  
    'loan_percent_income': 'ratio_pret_revenu',
    'cb_person_default_on_file': 'defaut_historique_credit',
    'cb_person_cred_hist_length': 'duree_historique_credit'
}


data = data_clean.rename(columns=colonnes_fr)



```

```python
from pandas.api.types import CategoricalDtype

print("Types de données après nettoyage :")

#  colonnes categorielles
colonnes_categorielles = ['statut_logement', 'motif_pret', 'defaut_historique_credit']
for col in colonnes_categorielles:
    data[col] = data[col].astype('category')

# Colonnes ordinales

ordre_risque = CategoricalDtype(categories=['A', 'B', 'C', 'D', 'E', 'F', 'G'], ordered=True)
data['niveau_de_risque'] = data['niveau_de_risque'].astype(ordre_risque)

# 3. Variable cible (binaire 0 ou 1)
data['statut_pret'] = data['statut_pret'].astype('int8')


print(data.dtypes)
```

```python


def explorer(data):
    print("Exploration des données")
    print("Nombre de lignes et de colonnes :", data.shape)
    print("Types de données :")
    print(data.dtypes)
    print("\nValeurs manquantes :")
    print(    
    data.isnull().sum())
    print("Pourcentage  des valeurs  manquantes par colonnes: ")
    print(data_clean.isnull().mean() * 100)
    
    print("Statistiques descriptives :")
    print(data.describe())
    print("\nCorrélations :")
    print(data.corr(numeric_only=True))

explorer(data)
```

```python


nb_doublons = data.duplicated().sum()
print(f"Nombre total de lignes en double : {nb_doublons}")

# 2 Afficher toutes les lignes concernées par un doublon
if nb_doublons > 0:
    # keep=False permet de voir toutes les copies (original + doublon(s))
    lignes_doublons = data[data.duplicated(keep=False)]
    
    # Trier par les principales colonnes pour les voir côte à côte
    lignes_doublons = lignes_doublons.sort_values(by=['age', 'revenu_annuel', 'montant_pret'])
    
    print("\n--- Aperçu des lignes dupliquées ---")
    print(lignes_doublons.head(10)) # Affiche les 10 premières lignes en double
else:
    print("Aucun doublon strict détecté dans le jeu de données.")
```

```python


#  Pourcentage de valeurs manquantes par colonne
print("Pourcentage de valeurs manquantes par colonne :\n")
print(data.isnull().mean() * 100)


 
```

```python
print("\n--- Remplissage des valeurs manquantes ---")

mediane_emp = data['anciennete_emploi'].median()
data['anciennete_emploi'] = data['anciennete_emploi'].fillna(mediane_emp)

# Le niveau_de_risque étant correlé avec le taux d'intérêt, 
# on peut utiliser la médiane du taux d'intérêt par niveau de risque
#  pour remplir les valeurs manquantes  

data['taux_interet'] = data['taux_interet'].fillna(
    data.groupby('niveau_de_risque')['taux_interet'].transform('median')
)

# Sécurité (au cas où il resterait un NaN résiduel) : on applique la médiane globale
data['taux_interet'] = data['taux_interet'].fillna(data['taux_interet'].median())


print("--- État des valeurs manquantes après traitement ---")
print(data.isnull().sum())
```

```python
import numpy as np
import pandas as pd

# On exclut la variable cible (statut_pret) et l'âge/historique si on ne veut pas les toucher
colonnes_a_traiter = ['revenu_annuel', 'anciennete_emploi', 'montant_pret', 'ratio_pret_revenu']

print("--- Plafonnement (Winsorisation) des colonnes continues ---")

for col in colonnes_a_traiter:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    
    borne_sup = Q3 + 1.5 * IQR
    
    # Compter et plafonner
    nb_modifies = (data[col] > borne_sup).sum()
    data.loc[data[col] > borne_sup, col] = borne_sup
    
    print(f"- {col} : {nb_modifies} valeurs plafonnées à {borne_sup:.2f}")
```
